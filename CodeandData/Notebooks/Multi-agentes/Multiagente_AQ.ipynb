```python


class Agent:
    """A simple class to represent an improvisational agent."""
    def __init__(self, name: str, persona: str):
        self.name = name
        self.persona = persona
        self.conversation_history = []

    def get_response(self, conversation_context: list):
        """Generates a response based on the persona and conversation history."""

        # We start with the agent's persona as a system message.
        messages = [
            {"role": "system", "content": self.persona}
        ]

        # Add the existing conversation history to the messages for context.
        messages.extend(conversation_context)

        # Add a prompt for the agent to take its turn.
        messages.append({"role": "user", "content": f"{self.name}, what do you say?"})

        try:
            response = client.chat.completions.create(
                model="gpt-4o", # You can use other models like "gpt-4o"
                messages=messages,
                #max_tokens=70,
                temperature=0.8, # A higher temperature encourages more creative and humorous responses.
                top_p=0.6,
            )
            return response.choices[0].message.content.strip()
        except Exception as e:
            return f"Error: {e}"

def run_improvisation(agents: list, scenario: str, turns: int):
    """
    Runs a humorous improvisation session between the provided agents.
    """
    print("--- STARTING IMPROVISATION ---")
    print(f"SCENARIO: {scenario}\n")

    # The conversation history is shared between all agents to maintain context.
    shared_history = [
        {"role": "system", "content": f""" Esta es una improvisación humorística en español mexicano. Todos los agentes deben improvisar una conversación y mantenerse en personaje,
        deben seguir la regla 'sí y además...' (yes-and) de de la improvisación teatral. Deben ser graciosos. Para esta improvisación, el agente taxista debe adivinar quién es el usuario que viaja con él, debe adivinar en los turnos finales.
        REGLA ESTRICTA: cada agente debe generar su línea de diálogo, no actuar como narrador ni generar diálogo de otros personajes.  El escenario es: {scenario}"""},
        {"role": "system", "content": f"""La improvisación por cada agente debe concretarse en el número asignado de tokens. Además, la idea debe concretarse en ese límite y debe ser verosímil. El escenario es: {scenario}"""}
    ]

    last_agent = None
    for i in range(turns):
        # Select an agent randomly, ensuring it's not the same as the last one
        eligible_agents = [agent for agent in agents if agent != last_agent]
        current_agent = random.choice(eligible_agents)
        last_agent = current_agent

        # The agent generates a new line based on the shared history.
        new_line = current_agent.get_response(shared_history)

        #print(f"{current_agent.name}: {new_line}")
        print(f"Turn {i+1}: {current_agent.name} says: {new_line}")

        # Add the new line to the shared history for the next turn.
        shared_history.append({"role": "assistant", "content": f"{current_agent.name}: {new_line}"})

        # A short delay to make it feel more like a live conversation.
        time.sleep(0.5)


    print("\n--- END OF IMPROVISATION ---")

if __name__ == "__main__":
    # Define the agents with their names and personas.
    agents = [
        Agent(
            name="Taxista",
            persona="Hombre mexicano de 53 años quien es Taxista de la Ciudad de México. Es una persona amable, conversadora, sociable, humorística e irónica"
        ),
        Agent(
            name="Dr.House",
            persona="Hombre estadounidense de 45 años quien es doctor, nefrólogo e infectólogo. Es una persona sarcástica, misántropa, seria y enojona"
        )
    ]

    # Define the scenario for the improv sketch.
    scenario = "Todo ocurre dentro de un taxi tradicional de la Ciudad de México, el chofer debe llevar al personaje a un pub"

    # Run the improv for a set number of turns.
    run_improvisation(agents, scenario, turns=12)
```
