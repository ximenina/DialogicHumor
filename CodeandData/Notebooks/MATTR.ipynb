## Moving Average Type-Token Ratio (MATTR)

MATTR addresses some limitations of the basic TTR by calculating the ratio over a moving window of text, providing a more stable and less sample-size-dependent measure of lexical diversity. It helps to overcome the issue where TTR tends to decrease as the text length increases. A common window size is 100 or 500 tokens.

Here's how to compute MATTR:

```python
import string
import re
from collections import Counter

file_path = '/content/c_espontaneo.txt'

with open(file_path, 'r', encoding='utf-8') as f:
  lines = f.readlines()


# Unir todas las líneas en un solo bloque para procesar más rápido
full_esp = "".join(lines).lower()

# Definir una limpieza extendida
full_esp = re.sub(r'[^\w\s]', ' ', full_esp)

# Tokenización
words_esp = full_esp.split()
###############
file_path = '/content/c_sintetico.txt'

with open(file_path, 'r', encoding='utf-8') as f:
  lines = f.readlines()


# Unir todas las líneas en un solo bloque para procesar más rápido
full_sin = "".join(lines).lower()

# Definir una limpieza extendida
full_sin = re.sub(r'[^\w\s]', ' ', full_sin)

# Tokenización
words_sin = full_sin.split()
```

```python
from collections import Counter

def calculate_ttr(word_list):
    if not word_list:
        return 0
    return len(set(word_list)) / len(word_list)

def calculate_mattr(words, window_size=100):
    if len(words) < window_size:
        return calculate_ttr(words) # Fallback to basic TTR if text is shorter than window

    ttr_values = []
    for i in range(len(words) - window_size + 1):
        window = words[i : i + window_size]
        ttr_values.append(calculate_ttr(window))

    return sum(ttr_values) / len(ttr_values)

# Ensure 'words' is available from previous steps
# If not, you might need to re-run the tokenization cell
# words = full_text.split()

mattr_value = calculate_mattr(words_esp, window_size=100)
print(f"Moving Average Type-Token Ratio (MATTR) with window size 100 (esp): {mattr_value:.4f}")

mattr_value = calculate_mattr(words_sin, window_size=100)
print(f"Moving Average Type-Token Ratio (MATTR) with window size 100 (sin): {mattr_value:.4f}")
print("####")

mattr_value = calculate_mattr(words_esp, window_size=50)
print(f"Moving Average Type-Token Ratio (MATTR) with window size 50 (esp): {mattr_value:.4f}")

mattr_value = calculate_mattr(words_sin, window_size=50)
print(f"Moving Average Type-Token Ratio (MATTR) with window size 50 (sin): {mattr_value:.4f}")

print("####")
mattr_value = calculate_mattr(words_esp, window_size=25)
print(f"Moving Average Type-Token Ratio (MATTR) with window size 25 (esp): {mattr_value:.4f}")

mattr_value = calculate_mattr(words_sin, window_size=25)
print(f"Moving Average Type-Token Ratio (MATTR) with window size 25 (sin): {mattr_value:.4f}")
print("####")
mattr_value = calculate_mattr(words_esp, window_size=150)
print(f"Moving Average Type-Token Ratio (MATTR) with window size 150 (esp): {mattr_value:.4f}")

mattr_value = calculate_mattr(words_sin, window_size=150)
print(f"Moving Average Type-Token Ratio (MATTR) with window size 150 (sin): {mattr_value:.4f}")
```
